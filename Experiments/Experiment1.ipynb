Task 7: Generate summary statistics using df.describe(); record mean, minimum and
maximum values of Age and Fare.

```python
summary_stats = df_train.describe()
display(summary_stats)

print(f"\nMean Age: {summary_stats.loc['mean', 'Age']:.2f}")
print(f"Minimum Age: {summary_stats.loc['min', 'Age']:.2f}")
print(f"Maximum Age: {summary_stats.loc['max', 'Age']:.2f}")

print(f"\nMean Fare: {summary_stats.loc['mean', 'Fare']:.2f}")
print(f"Minimum Fare: {summary_stats.loc['min', 'Fare']:.2f}")
print(f"Maximum Fare: {summary_stats.loc['max', 'Fare']:.2f}")
```

Task 8: Identify missing values using df.isnull().sum().

```python
missing_values = df_train.isnull().sum()
display(missing_values)
```

Task 9: Fill missing Age values with the median.

```python
median_age = df_train['Age'].median()
df_train['Age'].fillna(median_age, inplace=True)

# Verify that missing 'Age' values have been filled
print(f"Missing 'Age' values after filling: {df_train['Age'].isnull().sum()}")
```

Task 10: Filter surviving passengers (Survived == 1).

```python
df_survived = df_train[df_train['Survived'] == 1]
display(df_survived.head())
```

Task 11: Convert Fare column to a NumPy array.

```python
fare_array = df_train['Fare'].to_numpy()

print(f"Type of fare_array: {type(fare_array)}")
print(f"First 5 elements of fare_array: {fare_array[:5]}")
```

Task 12: Calculate maximum and average Fare using np.max() and np.mean().

```python
max_fare = np.max(fare_array)
mean_fare = np.mean(fare_array)

print(f"Maximum Fare: {max_fare:.2f}")
print(f"Average Fare: {mean_fare:.2f}")
```

Task 13: Plot histogram of Age.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_train['Age'], bins=20, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

Task 14: Plot bar chart of Survival Count.

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=df_train, palette='viridis')
plt.title('Survival Count (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.xticks(ticks=[0, 1], labels=['Not Survived', 'Survived'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

Task 15: Plot scatter graph of Age vs Fare.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=df_train, alpha=0.7)
plt.title('Scatter Plot of Age vs. Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
